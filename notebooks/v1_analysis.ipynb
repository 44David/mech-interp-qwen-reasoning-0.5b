##### Setting up and loading qwen-reasoning-0.5b-v1

```python
from transformer_lens import HookedTransformer
import circuitsvis as cv
from transformers import AutoModelForCausalLM, AutoTokenizer

device = "cpu" # running on my laptop, use cuda if you have a gpu
model_path = "44David/qwen-0.5b-reasoning-v1" # my model uploaded to hf

hf_model = AutoModelForCausalLM.from_pretrained(model_path,).to(device)

tokenizer = AutoTokenizer.from_pretrained(model_path,)

model = HookedTransformer.from_pretrained(
    "Qwen/Qwen2.5-0.5B",
    hf_model=hf_model,
    tokenizer=tokenizer,
    device=device,
    center_writing_weights=False,
    center_unembed=False,
)
```

##### Analyzing and visualizing attention patterns

```python
# gsm8k-style question
prompt = "A bakery makes 12 batches of cookies. Each batch has 24 cookies. They sell 3/4 of all the cookies. How many cookies are left?"

tokens = model.to_tokens(prompt)
logits, cache = model.run_with_cache(tokens)

attention_pattern = cache["pattern", 0]

str_tokens = model.to_str_tokens(tokens)

display(cv.attention.attention_patterns(
    tokens=str_tokens,
    attention=attention_pattern[0],
))

```
